# K近邻-分类算法
计算待分类样本与训练集样本的距离，距离最短的前$K$个训练集样本的标签，就认为是待测样本的标签。这里不考虑最高票数不唯一的情况

距离选用Euclidean距离，即2-平方范数
$$
d\left(\mathbf{x},\mathbf{y}\right)=\left\lbrack\sum_{k=1}^{n}\left(x_{k}-y_{k}\right)^2\right\rbrack^{\frac12}
$$
其中，
$$
\mathbf{x}=\begin{pmatrix}x_1\\ \cdots\\ x_{i}\\ \cdots\\ x_{n}\end{pmatrix},\mathbf{y}=\begin{pmatrix}y_1\\ \cdots\\ y_{i}\\ \cdots\\ y_{n}\end{pmatrix}
$$

---
有$N$个样本，且每个样本有$M$个特征，如下
|0|特征1|...|特征$j$|特征$j+1$|...|特征$M$|类别标签|
|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|
|样本1|...|...|...|...|...|...|...|
|...|...|...|...|...|...|...|...|
|样本$i$|$x_{i1}$|...|$x_{ij}$|$x_{i(j+1)}$|...|$x_{iM}$|$y_{i}$|
|...|...|...|...|...|...|...|...|
|样本$N$|...|...|...|...|...|...|...|
|待分类样本|...|...|...|...|...|...|未知|

已知样本的特征矩阵，已知样本的类别标签向量，待测样本的特征向量分别为
$$
\mathbf{X}=\begin{pmatrix}\cdots & \cdots & \cdots & \cdots & \cdots & \cdots\\ \cdots & \cdots & \cdots & \cdots & \cdots & \cdots\\ x_{i1} & \cdots & x_{ij} & x_{i\left(j+1\right)} & \cdots & x_{iM}\\ \cdots & \cdots & \cdots & \cdots & \cdots & \cdots\\ \cdots & \cdots & \cdots & \cdots & \cdots & \cdots\end{pmatrix}=\begin{pmatrix}\cdots\\ \cdots\\ \alpha_{i}\\ \cdots\\ \cdots\end{pmatrix},\mathbf{y}=\begin{pmatrix}\cdots\\ \cdots\\ y_{i}\\ \cdots\\ \cdots\end{pmatrix},\tilde{\mathbf{x}}=\begin{pmatrix}\cdots, & \cdots, & \tilde{x}_{j}, & \cdots, & \cdots\end{pmatrix}
$$
距离的平方为
$$
\begin{align*} & d^2\left(\tilde{\mathbf{x}},\alpha_{i}\right)=\sum_{j=1}^{M}\left(\tilde{x}_{j}-x_{ij}\right)^2=\begin{pmatrix}\tilde{x}_1-x_{i1}, & \cdots, & \tilde{x}_{j}-x_{ij}, & \cdots, & \tilde{x}_{M}-x_{iM}\end{pmatrix}\begin{pmatrix}\tilde{x}_1-x_{i1}\\ \cdots\\ \tilde{x}_{j}-x_{ij}\\ \cdots\\ \tilde{x}_{M}-x_{iM}\end{pmatrix}\\  & =\begin{pmatrix}\tilde{x}_1-x_{i1}\\ \cdots\\ \tilde{x}_{j}-x_{ij}\\ \cdots\\ \tilde{x}_{M}-x_{iM}\end{pmatrix}^{\mathrm{T}}\left[\begin{pmatrix}\tilde{x}_1\\ \cdots\\ \tilde{x}_{j}\\ \cdots\\ \tilde{x}_{M}\end{pmatrix}-\begin{pmatrix}x_{i1}\\ \cdots\\ x_{ij}\\ \cdots\\ x_{iM}\end{pmatrix}\right]=\left(\tilde{\mathbf{x}}^{\mathrm{T}}-\alpha_{i}^{\mathrm{T}}\right)^{\mathrm{T}}\left(\tilde{\mathbf{x}}^{\mathrm{T}}-\alpha_{i}^{\mathrm{T}}\right)=\left(\tilde{\mathbf{x}}-\alpha_{i}\right)\left(\tilde{\mathbf{x}}^{\mathrm{T}}-\alpha_{i}^{\mathrm{T}}\right)\\  & =\tilde{\mathbf{x}}\tilde{\mathbf{x}}^{\mathrm{T}}-\tilde{\mathbf{x}}\alpha_{i}^{\mathrm{T}}-\alpha_{i}\tilde{\mathbf{x}}^{\mathrm{T}}+\alpha_{i}\alpha_{i}^{\mathrm{T}}\\  & =\tilde{\mathbf{x}}\tilde{\mathbf{x}}^{\mathrm{T}}-2\alpha_{i}\tilde{\mathbf{x}}^{\mathrm{T}}+\alpha_{i}\alpha_{i}^{\mathrm{T}}\end{align*}
$$
拼成矩阵，得
$$
\begin{align*} & d^2\left(\tilde{\mathbf{x}},\mathbf{X}\right)=\begin{pmatrix}d^2\left(\tilde{\mathbf{x}},\alpha_1\right)\\ \cdots\\ d^2\left(\tilde{\mathbf{x}},\alpha_{i}\right)\\ \cdots\\ d^2\left(\tilde{\mathbf{x}},\alpha_{N}\right)\end{pmatrix}=\begin{pmatrix}\tilde{\mathbf{x}}\tilde{\mathbf{x}}^{\mathrm{T}}-2\alpha_1\tilde{\mathbf{x}}^{\mathrm{T}}+\alpha_1\alpha_1^{\mathrm{T}}\\ \cdots\\ \tilde{\mathbf{x}}\tilde{\mathbf{x}}^{\mathrm{T}}-2\alpha_{i}\tilde{\mathbf{x}}^{\mathrm{T}}+\alpha_{i}\alpha_{i}^{\mathrm{T}}\\ \cdots\\ \tilde{\mathbf{x}}\tilde{\mathbf{x}}^{\mathrm{T}}-2\alpha_{N}\tilde{\mathbf{x}}^{\mathrm{T}}+\alpha_{N}\alpha_{N}^{\mathrm{T}}\end{pmatrix}\\  & =\tilde{\mathbf{x}}\tilde{\mathbf{x}}^{\mathrm{T}}\begin{pmatrix}1\\ \cdots\\ 1\\ \cdots\\ 1\end{pmatrix}-2\begin{pmatrix}\alpha_1\tilde{\mathbf{x}}^{\mathrm{T}}\\ \cdots\\ \alpha_{i}\tilde{\mathbf{x}}^{\mathrm{T}}\\ \cdots\\ \alpha_{N}\tilde{\mathbf{x}}^{\mathrm{T}}\end{pmatrix}+\begin{pmatrix}\alpha_1\alpha_1^{\mathrm{T}}\\ \cdots\\ \alpha_{i}\alpha_{i}^{\mathrm{T}}\\ \cdots\\ \alpha_{N}\alpha_{N}^{\mathrm{T}}\end{pmatrix}\\  & =\tilde{\mathbf{x}}\tilde{\mathbf{x}}^{\mathrm{T}}\left(\mathbf{1}\right)_{N}-2\begin{pmatrix}\alpha_1\\ \cdots\\ \alpha_{i}\\ \cdots\\ \alpha_{N}\end{pmatrix}\tilde{\mathbf{x}}^{\mathrm{T}}+\begin{pmatrix}\alpha_1\alpha_1^{\mathrm{T}}\\ \cdots\\ \alpha_{i}\alpha_{i}^{\mathrm{T}}\\ \cdots\\ \alpha_{N}\alpha_{N}^{\mathrm{T}}\end{pmatrix}\\  & =\tilde{\mathbf{x}}\tilde{\mathbf{x}}^{\mathrm{T}}\left(\mathbf{1}\right)_{N}-2\mathbf{X}\tilde{\mathbf{x}}^{\mathrm{T}}+\begin{pmatrix}\alpha_1\alpha_1^{\mathrm{T}}\\ \cdots\\ \alpha_{i}\alpha_{i}^{\mathrm{T}}\\ \cdots\\ \alpha_{N}\alpha_{N}^{\mathrm{T}}\end{pmatrix}\end{align*}
$$
上面最后的结果，第$3$个向量也是很好求解的，如下
$$
\begin{align*} & \mathbf{XX}^{\mathrm{T}}=\begin{pmatrix}\alpha_1\\ \cdots\\ \alpha_{i}\\ \cdots\\ \alpha_{N}\end{pmatrix}\begin{pmatrix}\alpha_1^{\mathrm{T}}, & \cdots, & \alpha_{i}^{\mathrm{T}}, & \cdots, & \alpha_{N}^{\mathrm{T}}\end{pmatrix}\\  & =\begin{pmatrix}\alpha_1\alpha_1^{\mathrm{T}} & \cdots & \cdots & \cdots & \cdots\\ \cdots & \cdots & \cdots & \cdots & \cdots\\ \cdots & \cdots & \alpha_{i}\alpha_{i}^{\mathrm{T}} & \cdots & \cdots\\ \cdots & \cdots & \cdots & \cdots & \cdots\\ \cdots & \cdots & \cdots & \cdots & \alpha_{N}\alpha_{N}^{\mathrm{T}}\end{pmatrix}\end{align*}
$$
可以看出，所求的向量就是上述矩阵的主对角线的各元素

我们用$\mathrm{diagfrom}()$（`Python`中也有这种功能的函数，不过我们不用那个写法，这里选择自创一个）表示矩阵的主对角线元素所组成的列向量，即
$$
\mathrm{diagfrom}\begin{pmatrix}\alpha_1\alpha_1^{\mathrm{T}} & \cdots & \cdots & \cdots & \cdots\\ \cdots & \cdots & \cdots & \cdots & \cdots\\ \cdots & \cdots & \alpha_{i}\alpha_{i}^{\mathrm{T}} & \cdots & \cdots\\ \cdots & \cdots & \cdots & \cdots & \cdots\\ \cdots & \cdots & \cdots & \cdots & \alpha_{N}\alpha_{N}^{\mathrm{T}}\end{pmatrix}=\begin{pmatrix}\alpha_1\alpha_1^{\mathrm{T}}\\ \cdots\\ \alpha_{i}\alpha_{i}^{\mathrm{T}}\\ \cdots\\ \alpha_{N}\alpha_{N}^{\mathrm{T}}\end{pmatrix}
$$
则最终结果可表示为
$$
d^2\left(\tilde{\mathbf{x}},\mathbf{X}\right)=\tilde{\mathbf{x}}\tilde{\mathbf{x}}^{\mathrm{T}}\left(\mathbf{1}\right)_{N}-2\mathbf{X}\tilde{\mathbf{x}}^{\mathrm{T}}+\mathrm{diagfrom}\left(\mathbf{XX}^{\mathrm{T}}\right)
$$

---
现在我们用以个简单的例子计算一下，最近邻数选择3
|0|特征1|特征2|类别标签|
|:---:|:---:|:---:|:---:|
|样本1|1|2|0|
|样本2|2|3|0|
|样本3|3|3|0|
|样本4|6|5|1|
|样本5|7|7|1|
|样本6|8|6|1|
|---|---|---|---|
|待分类样本|4|4|未知|

```python
# 导入库
import numpy as np
```

```python
# 获取数据
X = np.array([
    [1, 2],
    [2, 3],
    [3, 3],
    [6, 5],
    [7, 7],
    [8, 6]
])
y = np.array([
    [0],
    [0],
    [0],
    [1],
    [1],
    [1]
])
x_tilde = np.array([4, 4])

N = len(y)    # 样本个数

# 打印部分数据
print(f"样本的个数：{N}")
print(f"X的形状：{X.shape}")
print(f"y的形状：{y.shape}")
print(f"x_tilde的形状：{x_tilde.shape}")
```

```python
# 计算距离
beta0 = (x_tilde @ x_tilde.T) * np.ones((N, 1))    # 上面公式的第1个向量，不带系数
beta1 = X @ (x_tilde.T).reshape(-1, 1)    # 上面公式的第2个向量，不带系数。算出来是行向量，应用.reshape()转换成列向量
beta2 = (np.diagonal(X @ X.T)).reshape(-1, 1)    # 上面公式的第3个向量，不带系数。算出来是行向量，应用.reshape()转换成列向量

distances = np.sqrt(beta0 - 2 * beta1 + beta2)

# 打印部分数据
print(f"3个向量依次为：\n{beta0}\n{beta1}\n{beta2}")
print(f"距离分别为：\n{distances}")
```

```python
# 将得出的距离与对应的类别标签组合起来。在这里选用列表，其元素为元组
results = [(distances[i][0], y[i][0]) for i in np.arange(N)]    # 后面都加上“[0]”，是因为distances[i]为(array([3.60555128])，我们需要的是里面的数据

# 打印部分数据
print(f"组装好后的列表：\n{results}")
print(f"列表第1个元素：{results[0]}")    # 第1个元组
print(f"列表第1个元素的第1个元素：{results[0][0]}")    # 第1个元组的第1个元素
```

```python
# 根据距离，对列表的元组排序
sorted_results = sorted(    # 排序，返回给新变量
    results,    # 要排序的变量
    key=lambda results: results[0],    # 排序依据，元组的第1个元素。注意，这一行的“results"仅作为定义的变量，完全可以用其他形式的变量
    reverse=False    # 不是倒序排列，也就是按正序（这行不屑也可以，默认为False）
)

# 打印部分数据
print(sorted_results)
```

```python
# 输出预测结果
n_neighbors = 3    # 最近邻的数量
y_neighbors = {}    # 初始化，用于装填前3个类别标签

for i in sorted_results[:n_neighbors]:    # (np.float64(1.4142135623730951), np.int64(0))...，需要取其中的第2个元素。只需要前3个元组
    i = i[1]    # 0,0,1
    y_neighbors[i] = y_neighbors.get(i, 0) + 1
"""
    dictionary.get(key, 0)表示获取字典dictionary的键key对应的值value，若不存在该键key，则返回0
    上面代码的含义为：
        第1次，y_neighbors为空字典，没有键值对。获取该字典的键0对应的值，因为不存在，则返回0，再将最终结果返回给字典的键，也就是创建键值对：0:1
            y_neighbors = {0: 1}
        第2次获取字典的键0对应的值。已存在该键，故得到1，再+1并将结果返回给字典的键0
            y_neighbors = {0: 2}
        第3次获取字典的键1对应的值。该键不存在，故得到0，再+1并将结果返回给字典的键1
            y_neighbors = {
                0: 2,
                1: 1
            }
"""

y_tilde = max(y_neighbors, key=y_neighbors.get)

# 打印部分数据
print(f"组装完成的标签字典为：{y_neighbors}")
print(f"预测值为：{y_tilde}")
```
