## 逻辑回归
研究的标签类别为$0,1$，也就是取值为$y_{i}=0,1$。有$N$个样本且每个样本有$M$个特征

先推导出损失函数表达式，再求出损失函数的梯度，然后根据梯度下降法迭代更新公式

---
先用一组线性关系来得到一组预测值，即
$$
\begin{align*} & z_{i}=w_1x_{i1}+\cdots+w_{j}x_{ij}+\cdots+w_{M}x_{iM}+b\\  & =\begin{pmatrix}x_{i1}, & \cdots, & x_{ij}, & \cdots, & x_{iM}\end{pmatrix}\begin{pmatrix}w_1\\ \cdots\\ w_{j}\\ \cdots\\ x_{iM}\end{pmatrix}+b=\mathbf{x}_{i}^{\mathrm{T}}\mathbf{w}+b\end{align*}
$$
Sigmoid函数为
$$
\sigma\left(z\right)=\frac{1}{1+\mathrm{e}^{-z}}
$$
这里给出Sigmoid函数的导数的一个特点，
$$
\begin{align*} & \frac{\mathrm{d}}{\mathrm{dz}}\sigma\left(z\right)=\left(-1\right)\cdot\left(\frac{1}{1+\mathrm{e}^{-z}}\right)^2\cdot\left(-\mathrm{e}^{-z}\right)=\frac{1}{1+\mathrm{e}^{-z}}\cdot\frac{\mathrm{e}^{-z}}{1+\mathrm{e}^{-z}}\\  & =\sigma\left(z\right)\cdot\frac{\left(1+\mathrm{e}^{-z}\right)-1}{1+\mathrm{e}^{-z}}=\sigma\left(z\right)\cdot\left[1-\sigma\left(z\right)\right]\end{align*}
$$
将得到的预测值$z_{i}$代入Sigmoid函数，映射为一组取值在$(0,1)$的数

我们用$p_{i}$表示预测值所映射成的值，即$p_{i}=\sigma\left(z_{i}\right)=\mathbf{x}_{i}^{\mathrm{T}}\mathbf{w}+b$

同时假设$\sigma\left(z_{i}\right)$表示标签类别为$y_{i}=1$的概率
这应该理解为条件概率，是$y_{i}=1$在“样本为$\mathbf{x}_{i}$”的条件下发生的，故有
$$
P\left(Y_{i}=1\left|\mathbf{x}_{i}\right.\right)=\sigma\left(z_{i}\right)=p_{i}
$$
那么类别为$y_{i}=0$的概率为
$$
P\left(Y_{i}=0\left|\mathbf{x}_{i}\right.\right)=1-p_{i}
$$
将上面两种情况写成紧凑形式，有
$$
P\left(Y_{i}=y_{i}\left|\mathbf{x}_{i}\right.\right)=p_{i}^{y_{i}}\left(1-p_{i}\right)^{1-y_{i}},y_{i}=0,1
$$
这就是概率统计中的$0-1$分布

实际有$N$个样本，我们认为这些样本是相互独立的。多个相互独立的随机变量，其同时发生的概率记为各自发生概率的乘积

当这些样本同时发生时（也必须同时发生，因为我们要同时处理这些样本），其概率称为极大似然函数，有
$$
L\left(w,b\right)=\prod_{i=1}^{N}P\left(Y_{i}=y_{i}\left|\mathbf{x}_{i}\right.\right)=\prod_{i=1}^{N}p_{i}^{y_{i}}\left(1-p_{i}\right)^{1-y_{i}}
$$
这个函数应当存在极大值，再次不作证明

按理说，后续对极大似然求极值，对应的极大值点即为最优参数，这称为极大似然估计。事实上不会直接求偏导的

极大似然估计很好理解，在样本为$\mathbf{x}_{i}$的条件下，发生了$(y_1,\cdots,y_{N})$，既然这些类别发生了，我们就认为这种情况是在所有可能的情况中，处于极大值的位置

由于函数$\ln{x}$对于变量$x$是单调增的，可以得知，$\ln{f(x)}$与$f(x)$有相同的单调性，那么两者的极值点也就相同

先给出对数运算的公式
$$
\begin{align*} & \ln{\prod_{i=1}^{N}a_{i}^{b_{i}}}=\ln\left(a_1^{b_1}\cdots a_{i}^{b_{i}}\cdots a_{N}^{b_{N}}\right)=b_1\ln a_1+\cdots+b_{i}\ln a_{i}+\cdots+b_{N}\ln a_{N}\\  & =\sum_{i=1}^{N}b_{i}\ln a_{i}\end{align*}
$$
极大似然函数直接求偏导并不方便，而当其取对数后，求导很是简单，极大似然对数函数为
$$
\ln{L}\left(\mathbf{w},b\right)=\sum_{i=1}^{N}\left[y_{i}\ln p_{i}+\left(1-y_{i}\right)\ln\left(1-y_{i}\right)\right]
$$
讨论最优参数，应该是用损失函数来分析

损失函数选择交叉熵损失函数，表达式为
> $$
J\left(\mathbf{w},b\right)=-\frac{1}{N}\ln{L}\left(\mathbf{w},b\right)=-\frac{1}{N}\sum_{i=1}^{N}\left[y_{i}\ln p_{i}+\left(1-y_{i}\right)\ln\left(1-y_{i}\right)\right]
$$
其中，$p_{i}=\sigma\left(\mathbf{x}_{i}^{\mathrm{T}}\mathbf{w}+b\right)$

因为极大似然对数函数存在极大值，那么交叉熵损失函数存在极小值

---
这里用梯度下降法来迭代参数$\mathbf{w}$和$b$

对于$z_{i}=w_1x_{i1}+\cdots+w_{j}x_{ij}+\cdots+w_{M}x_{iM}+b$，记$x_{i(M+1)}=1,w_{M+1}=b$，得
$$
\begin{align*} & z_{i}=w_1x_{i1}+\cdots+w_{j}x_{ij}+\cdots+w_{M}x_{iM}+w_{\left(M+1\right)}x_{i\left(M+1\right)}\\  & =\begin{pmatrix}x_{i1}, & \cdots, & x_{ij}, & \cdots, & x_{i\left(M+1\right)}\end{pmatrix}\begin{pmatrix}w_1\\ \cdots\\ w_{j}\\ \cdots\\ x_{i\left(M+1\right)}\end{pmatrix}=\mathbf{x}_{i}^{\mathrm{T}}\mathbf{w}\end{align*}
$$
注意这里的$\mathbf{x}_{i},\mathbf{w}$与前面的不同

于是
$$
J\left(\mathbf{w}\right)=-\frac{1}{N}\ln{L}\left(\mathbf{w},b\right)=-\frac{1}{N}\sum_{i=1}^{N}\left[y_{i}\ln p_{i}+\left(1-y_{i}\right)\ln\left(1-p_{i}\right)\right]
$$
其中，$p_{i}=\sigma\left(z_{i}\right),z_{i}=\mathbf{x}_{i}^{\mathrm{T}}\mathbf{w}$

这是一个复合函数，需要用链式求导法则，先求出内层函数的导数
$$
\begin{align*} & \nabla_{\mathbf{w}}z_{i}=\mathbf{x}_{i}\\  & \frac{\partial p_{i}}{\partial z_{i}}=\sigma\left(z_{i}\right)\left[1-\sigma\left(z_{i}\right)\right]=p_{i}\left(1-p_{i}\right)\\  & \begin{align*}\frac{\partial}{\partial p_{i}} & \left[y_{i}\ln p_{i}+\left(1-y_{i}\right)\ln\left(1-y_{i}\right)\right]=\left\lbrack y_{i}\cdot\frac{1}{p_{i}}+\left(1-y_{i}\right)\cdot\frac{-1}{1-p_{i}}\right\rbrack=\frac{y_{i}-p_{i}}{p_{i}\left(1-p_{i}\right)}\end{align*}\end{align*}
$$
其中用到的公式分别在Lasso回归中，和本节前面的Sigmoid函数

在下面，从“$\doteq$”开始，将$x_{M+1}=1$从$\mathbf{x}$中分离，得
$$
\begin{align*} & \nabla_{\mathbf{w}}J\left(\mathbf{w}\right)=-\frac{1}{N}\sum_{i=1}^{N}\frac{\partial}{\partial p_{i}}\left[y_{i}\ln{p_{i}}-\left(1-y_{i}\right)\ln{\left(1-p_{i}\right)}\right]\cdot\frac{\partial p_{i}}{\partial z_{i}}\cdot\nabla_{\mathbf{w}}z_{i}\\  & =-\frac{1}{N}\sum_{i=1}^{N}\frac{y_{i}-p_{i}}{p_{i}\left(1-p_{i}\right)}\cdot p_{i}\left(1-p_{i}\right)\cdot\mathbf{x}_{i}^{\mathrm{T}}=\frac{1}{N}\sum_{i=1}^{N}\left(p_{i}-y_{i}\right)\mathbf{x}_{i}\\  & =\frac{1}{N}\sum_{i=1}^{N}\left(p_{i}-y_{i}\right)\begin{pmatrix}x_{i1}\\ \cdots\\ x_{ij}\\ \cdots\\ x_{iM}\\ x_{i\left(M+1\right)}\end{pmatrix}\doteq\frac{1}{N}\sum_{i=1}^{N}\left(p_{i}-y_{i}\right)\begin{pmatrix}\hat{\mathbf{x}}_{i}\\ 1\end{pmatrix}\\  & =\frac{1}{N}\sum_{i=1}^{N}\left(p_{i}-y_{i}\right)=\frac{1}{N}\begin{pmatrix}\sum\left(p_{i}-y_{i}\right)\hat{\mathbf{x}}_{i}\\ \sum\left(p_{i}-y_{i}\right)1\end{pmatrix}\\  & =\frac{1}{N}\begin{pmatrix}\begin{pmatrix}\cdots, & \hat{\mathbf{x}}_{N}\end{pmatrix}\begin{pmatrix}\cdots\\ p_{N}-y_{N}\end{pmatrix}\\ \sum\left(p_{i}-y_{i}\right)1\end{pmatrix}=\frac{1}{N}\begin{pmatrix}\hat{\mathbf{X}}^{\mathrm{T}}\left(\mathbf{p}-\mathbf{y}\right)\\ \sum\left(p_{i}-y_{i}\right)\end{pmatrix}\end{align*}
$$
现在我们需要分离出权重$\mathbf{w}$和截距$b$，从现在开始，权重$\mathbf{w}$不在含有$b$，
$$
\begin{align*} & \nabla_{\mathbf{w}}J\left(\mathbf{w},b\right)=\frac{1}{N}\mathbf{X}^{\mathrm{T}}\left(\mathbf{p}-\mathbf{y}\right)\\  & \frac{\partial}{\partial b}J\left(\mathbf{w},b\right)=\frac{1}{N}\sum_{i=1}^{N}\left(p_{i}-y_{i}\right)\end{align*}
$$
因此，可以根据梯度下降法迭代权重$\mathbf{w}$和截距$b$的更新公式，有
> $$
\begin{align*} & \mathbf{w}^{\left(k+1\right)}=\mathbf{w}^{\left(k\right)}-\eta\frac{1}{N}\mathbf{X}^{\mathrm{T}}\left(\mathbf{p}^{\left(k\right)}-\mathbf{y}\right)\\  & b^{\left(k+1\right)}=b^{\left(k\right)}-\eta\frac{1}{N}\sum_{i=1}^{N}\left(p_{i}^{\left(k\right)}-y_{i}\right)\end{align*}
$$
