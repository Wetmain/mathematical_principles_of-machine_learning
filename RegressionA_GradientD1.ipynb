## 梯度下降法-多元线性回归
给出推导过程涉及的矩阵运算公式，涉及到常见用法的逆向过程

**a）** 向量加法：
$$
\alpha=\begin{pmatrix}a_1\\ \cdots\\ a_{n}\end{pmatrix},\beta=\begin{pmatrix}b_1\\ \cdots\\ b_{n}\end{pmatrix}
$$
常用
$$
\alpha+\beta=\begin{pmatrix}a_1\\ \cdots\\ a_{n}\end{pmatrix}+\begin{pmatrix}b_1\\ \cdots\\ b_{n}\end{pmatrix}=\begin{pmatrix}a_1+b_1\\ \cdots\\ a_{n}+b_{n}\end{pmatrix}
$$
在这里
> $$
\begin{pmatrix}a_1+b_1\\ \cdots\\ a_{n}+b_{n}\end{pmatrix}=\begin{pmatrix}a_1\\ \cdots\\ a_{n}\end{pmatrix}+\begin{pmatrix}b_1\\ \cdots\\ b_{n}\end{pmatrix}=\alpha+\beta
$$
**b）** 向量内积：
$$
\alpha=\begin{pmatrix}a_1\\ \cdots\\ a_{n}\end{pmatrix},\beta=\begin{pmatrix}b_1\\ \cdots\\ b_{n}\end{pmatrix}
$$
常用
$$
\alpha^{\mathrm{T}}\beta=\begin{pmatrix}a_1 & \cdots & a_{n}\end{pmatrix}\begin{pmatrix}b_1\\ \cdots\\ b_{n}\end{pmatrix}=a_1b_1+\cdots+a_{n}b_{n}
$$
在这里
> $$
a_1b_1+\cdots+a_{n}b_{n}=\begin{pmatrix}a_1 & \cdots & a_{n}\end{pmatrix}\begin{pmatrix}b_1\\ \cdots\\ b_{n}\end{pmatrix}=\alpha^{\mathrm{T}}\beta
$$
**c）** 分块矩阵：
将矩阵切成一堆行向量
$$
\mathbf{A}=\begin{pmatrix}\alpha_1\\ \cdots\\ \alpha_{n}\end{pmatrix}
$$
常用
$$
\mathbf{A}\gamma=\begin{pmatrix}\alpha_1\\ \cdots\\ \alpha_{n}\end{pmatrix}\gamma=\begin{pmatrix}\alpha_1\gamma\\ \cdots\\ \alpha_{n}\gamma\end{pmatrix}
$$
在这里
> $$
\begin{pmatrix}\alpha_1\gamma\\ \cdots\\ \alpha_{n}\gamma\end{pmatrix}=\begin{pmatrix}\alpha_1\\ \cdots\\ \alpha_{n}\end{pmatrix}\gamma=\mathbf{A}\gamma
$$

---
现在推导估计值$\hat{y}$的矩阵形式的表达式，有$N$个样本且每个样本有$M$个特征，最终拟合出的函数形如
> $$
\hat{y} =\sum_{j=1}^{M}w_{j}x_{j}+b=\left(w_1x_1+\cdots+w_{M}x_{M}\right)+b
$$
用矩阵表示，有
$$
\hat{y} =\alpha\mathbf{w}+b
$$
其中，$\alpha=\left(x_{1},\cdots,x_{M}\right)$，$\mathbf{w}=\begin{pmatrix}w_1\\ \cdots\\ w_{i}\\ \cdots\\ w_{N}\end{pmatrix}$

设数据集的特征有$M$个，样本有$N$个，形如
|0|特征1|...|特征$j$|特征$j+1$|...|特征$M$|标签|
|:---:|:---:|:---:|:---:|:---:|:---:|:---:|:---:|
|样本1|...|...|...|...|...|...|...|
|...|...|...|...|...|...|...|...|
|样本$i$|$x_{i1}$|...|$x_{ij}$|$x_{i(j+1)}$|...|$x_{iM}$|$y_{i}$|
|...|...|...|...|...|...|...|...|
|样本$M$|...|...|...|...|...|...|...|

将全部样本的特征所组成的数表用矩阵表示，有
$$
\mathbf{X}=\begin{pmatrix}\cdots & \cdots & \cdots & \cdots & \cdots & \cdots\\ \cdots & \cdots & \cdots & \cdots & \cdots & \cdots\\ x_{i1} & \cdots & x_{ij} & x_{i\left(j+1\right)} & \cdots & x_{iM}\\ \cdots & \cdots & \cdots & \cdots & \cdots & \cdots\\ \cdots & \cdots & \cdots & \cdots & \cdots & \cdots\end{pmatrix}=\begin{pmatrix}\cdots\\ \cdots\\ \alpha_{i}\\ \cdots\\ \cdots\end{pmatrix}
$$

标签可表示为
$$
\mathbf{y}=\begin{pmatrix}\cdots\\ \cdots\\ y_{i}\\ \cdots\\ \cdots\end{pmatrix}
$$

在拟合时，对于第$i$个样本，应有
$$
\hat{\mathbf{y}_{i}}=\alpha_{i}\mathbf{w}+b
$$
其中，$\alpha=\left(x_{i1},\cdots,x_{iM}\right)$

对于所有标签组成的矩阵，有
$$
\hat{\mathbf{y}}=\begin{pmatrix}\cdots\\ \cdots\\ y_{i}\\ \cdots\\ \cdots\end{pmatrix}=\begin{pmatrix}\alpha_{1}\mathbf{w}+b\\ \cdots\\ \alpha_{i}\mathbf{w}+b\\ \cdots\\ \alpha_{N}\mathbf{w}+b\end{pmatrix}
$$
根据矩阵的运算，上式又可表示为
$$
\hat{\mathbf{y}}=\begin{pmatrix}\alpha_1\mathbf{w}\\ \cdots\\ \alpha_i\mathbf{w}\\ \cdots\\ \alpha_N\mathbf{w}\end{pmatrix}+\begin{pmatrix}b\\ \cdots\\ b\\ \cdots\\ b\end{pmatrix}=\begin{pmatrix}\alpha_1\\ \cdots\\ \alpha_{i}\\ \cdots\\ \alpha_{N}\end{pmatrix}\mathbf{w}+\begin{pmatrix}b\\ \cdots\\ b\\ \cdots\\ b\end{pmatrix}={X}\mathbf{w}+\mathbf{b}
$$
其中，$\mathbf{b}=\begin{pmatrix}b\\ \cdots\\ b\\ \cdots\\ b\end{pmatrix}$

---
现在推导权重,截距的表达式

矩阵$\mathbf{X}$可以表示为
$$
\mathbf{X}=\begin{pmatrix}\beta_1^{\mathrm{T}}, & \cdots, & \beta_{j}^{\mathrm{T}}, & \cdots, & \beta_{M}^{\mathrm{T}}\end{pmatrix}
$$
其中
$$
\beta_{j}=\begin{pmatrix}x_{1j}, & \cdots, & x_{ij}, & \cdots, & x_{Nj}\end{pmatrix}
$$
可知
$$
\begin{pmatrix}\bm{\beta}_1\\ \cdots\\ \bm{\beta}_{j}\\ \cdots\\ \bm{\beta}_{M}\end{pmatrix}=\mathbf{X}^{\mathrm{T}}
$$

均方误差为
$$
J\left(\mathbf{w}^{\mathrm{T}},b\right)=\frac{1}{N}\sum_{i=1}^{N}\left(\hat{y}_{i}-y_{i}\right)^2=\frac{1}{N}\sum_{i=1}^{N}\left(w_1x_{i1}+\cdots+w_{j}x_{ij}+\cdots+w_{M}x_{iM}+b-y_{i}\right)^2
$$
其中，$J\left(\mathbf{w}^{\mathrm{T}},b\right)$也是损失函数

求偏导数，有
$$
\begin{align*} & \frac{\partial}{\partial w_{j}}J\left(\mathbf{w}^{\mathrm{T}},b\right)=\frac{2}{N}\sum_{i=1}^{N}\left(w_1x_{i1}+\cdots+w_{j}x_{ij}+\cdots+w_{M}x_{iM}+b-y_{i}\right)x_{ij}\\  & \frac{\partial}{\partial b}J\left(\mathbf{w}^{\mathrm{T}},b\right)=\frac{2}{N}\sum_{i=1}^{N}\left(w_1x_{i1}+\cdots+w_{j}x_{ij}+\cdots+w_{M}x_{iM}+b-y_{i}\right)\end{align*}
$$
对第一个式子的求和项，可简记为
$$
\sum_{i=1}^{N}\left(w_1x_{i1}+\cdots+w_{j}x_{ij}+\cdots+w_{M}x_{iM}+b-y_{i}\right)x_{ij}=\sum_{i=1}^{N}A_{i}x_{ij}=A_1x_{1j}+\cdots+A_{i}x_{ij}+\cdots+A_{N}x_{Nj}
$$
其中，
$$
A_{i}=w_1x_{i1}+\cdots+w_{j}x_{ij}+\cdots+w_{M}x_{iM}+b-y_{i}=\alpha_{i}\mathbf{w}+b-y_{i}
$$

进一步，用矩阵表示，得
$$
A_1x_{1j}+\cdots+A_{i}x_{ij}+\cdots+A_{N}x_{Nj}=\left(x_{1j},\cdots{},x_{Nj}\right)\begin{pmatrix}A_1\\ \cdots\\ A_{N}\end{pmatrix}=\mathbf{\beta}_{j}\mathbf{a}
$$
其中，
$$
\mathbf{a}=\begin{pmatrix}A_1\\ \cdots\\ A_{i}\\ \cdots\\ A_{N}\end{pmatrix}=\begin{pmatrix}\alpha_1\mathbf{w}+b-y_1\\ \cdots\\ \alpha_{i}\mathbf{w}+b-y_{i}\\ \cdots\\ \alpha_{N}\mathbf{w}+b-y_{N}\end{pmatrix}=\begin{pmatrix}\alpha_1\mathbf{w}\\ \cdots\\ \alpha_{i}\mathbf{w}\\ \cdots\\ \alpha_{N}\mathbf{w}\end{pmatrix}+\begin{pmatrix}b\\ \cdots\\ b\\ \cdots\\ b\end{pmatrix}-\begin{pmatrix}y_1\\ \cdots\\ y_{i}\\ \cdots\\ y_{N}\end{pmatrix}
$$
拼成矩阵，得
$$
\frac{\partial}{\partial\mathbf{w}}J\left(\mathbf{w}^{\mathrm{T}},b\right)=\frac{2}{N}\begin{pmatrix}\bm{\beta}_1\mathbf{a}\\ \cdots\\ \bm{\beta}_{j}\mathbf{a}\\ \cdots\\ \bm{\beta}_{M}\mathbf{a}\end{pmatrix}=\frac{2}{N}\begin{pmatrix}\bm{\beta}_1\\ \cdots\\ \bm{\beta}_{j}\\ \cdots\\ \bm{\beta}_{M}\end{pmatrix}\left[\begin{pmatrix}\alpha_1\mathbf{w}\\ \cdots\\ \alpha_{i}\mathbf{w}\\ \cdots\\ \alpha_{N}\mathbf{w}\end{pmatrix}+\begin{pmatrix}b\\ \cdots\\ b\\ \cdots\\ b\end{pmatrix}-\begin{pmatrix}y_1\\ \cdots\\ y_{i}\\ \cdots\\ y_{N}\end{pmatrix}\right]
$$
$$
=\frac{2}{N}\begin{pmatrix}\bm{\beta}_1\\ \cdots\\ \bm{\beta}_{j}\\ \cdots\\ \bm{\beta}_{M}\end{pmatrix}\left[\begin{pmatrix}\alpha_1\\ \cdots\\ \alpha_{i}\\ \cdots\\ \alpha_{N}\end{pmatrix}\mathbf{w}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right]=\frac{2}{N}\mathbf{X}^{\mathrm{T}}\left[\mathbf{Xw}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right]
$$
其中，
$$
(\mathbf{1})_{N}=\begin{pmatrix}1\\ \cdots\\ 1\\ \cdots\\ 1\end{pmatrix}_{N\times1},\mathbf{y}=\begin{pmatrix}y_1\\ \cdots\\ y_{i}\\ \cdots\\ y_{N}\end{pmatrix}_{N\times1}
$$
对于第二个式子得求和项，同理有
$$
\sum_{i=1}^{N}\left(w_1x_{i1}+\cdots+w_{j}x_{ij}+\cdots+w_{M}x_{iM}+b-y_{i}\right)=\sum_{i=1}^{N}A_{i}=A_1+\cdots+A_{i}+\cdots+A_{N}
$$
$$
=\left(\mathbf{1}\right)_{N}^{\mathrm{T}}\begin{pmatrix}A_1\\ \cdots\\ A_{i}\\ \cdots\\ A_{N}\end{pmatrix}=\left(\mathbf{1}\right)_{N}^{\mathrm{T}}\begin{pmatrix}\alpha_1\mathbf{w}+b-y_1\\ \cdots\\ \alpha_{i}\mathbf{w}+b-y_{i}\\ \cdots\\ \alpha_{N}\mathbf{w}+b-y_{N}\end{pmatrix}=\left(\mathbf{1}\right)_{N}^{\mathrm{T}}\left[\begin{pmatrix}\alpha_1\mathbf{w}\\ \cdots\\ \alpha_{i}\mathbf{w}\\ \cdots\\ \alpha_{N}\mathbf{w}\end{pmatrix}+\begin{pmatrix}b\\ \cdots\\ b\\ \cdots\\ b\end{pmatrix}-\begin{pmatrix}y_1\\ \cdots\\ y_{i}\\ \cdots\\ y_{N}\end{pmatrix}\right]
$$
得
$$
\frac{\partial}{\partial b}J\left(\mathbf{w}^{\mathrm{T}},b\right)=\frac{2}{N}\left(\mathbf{1}\right)_{N}^{\mathrm{T}}\left[\begin{pmatrix}\alpha_1\mathbf{w}\\ \cdots\\ \alpha_{i}\mathbf{w}\\ \cdots\\ \alpha_{N}\mathbf{w}\end{pmatrix}+\begin{pmatrix}b\\ \cdots\\ b\\ \cdots\\ b\end{pmatrix}-\begin{pmatrix}y_1\\ \cdots\\ y_{i}\\ \cdots\\ y_{N}\end{pmatrix}\right]=\frac{2}{N}\left(\mathbf{1}\right)_{N}^{\mathrm{T}}\left[\mathbf{Xw}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right]
$$


得梯度下降迭代公式为
> $$
\begin{align*} & \mathbf{w}^{\left(k+1\right)}=\mathbf{w}^{\left(k\right)}-\eta\frac{\partial}{\partial\mathbf{w}^{\left(k\right)}}J\left(\mathbf{w}^{\mathrm{T}\left(k\right)},b^{\left(k\right)}\right)=\mathbf{w}^{\left(k\right)}-\eta\frac{2}{N}\mathbf{X}^{\mathrm{T}}\left[\mathbf{Xw}^{\left(k\right)}+b^{\left(k\right)}\left(\mathbf{1}\right)_{N}-\mathbf{y}\right]\\  & b^{\left(k+1\right)}=b^{\left(k\right)}-\eta\frac{\partial}{\partial b^{\left(k\right)}}J\left(\mathbf{w}^{\mathrm{T}\left(k\right)},b^{\left(k\right)}\right)=\mathbf{b}^{\left(k\right)}-\eta\frac{2}{N}(\mathbf{1})_{N}^{\mathrm{T}}\left[\mathbf{Xw}^{\left(k\right)}+b^{\left(k\right)}\left(\mathbf{1}\right)_{N}-\mathbf{y}\right]\end{align*}
$$

---
取函数表达式为
$$
y=3x_1+4x_2+5x_3+2
$$
数据集的表达式为$y=3x_1+4x_2+5x_3+2+noise$

```python
# 导入库
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 创建随机数据集
random_speed = 2    # 指定随机种子
rng = np.random.default_rng(random_speed)    # 创建随机种子生成器

number_data = 150    # 样本总数
number_feature = 3    # 每个样本的特征数

X = np.zeros((number_data, number_feature))    # 初始化，150*3
for i in np.arange(number_data):    # 特征矩阵的行，0,1,2,...,148,149
    X[i] = [j for j in rng.uniform(-5, 5, number_feature)]    # 特征矩阵的列，0,1,2。这是列表推导式

noise = rng.normal(0, 1, 150)    # 生成噪声，1*150。正态分布，均值0，标准差1

y = np.zeros(number_data)    # 初始化，1*150
for i in np.arange(number_data):    # 0,1,2,...,148,149
    y[i] = 3 * X[i, 0] + 4 * X[i, 1] + 5 * X[i, 2] + noise[i]
y = y.reshape(-1, 1)    # 将行向量转换成列向量

# 打印部分数据
print(f"特征矩阵的前5行：\n{X[:5, :]}")
print(f"标签矩阵的前5行：\n{y[:5]}")
print(f"特征矩阵的形状：{X.shape}\n标签矩阵的形状：{y.shape}")
```

```python
# 划分训练集,测试集
rng1 = np.random.default_rng(random_speed+1)    # 更换一个随机种子生成器
random_numbers = rng1.permutation(number_data)    # 将序列[0,1,2,...,148,149]的顺序打乱

test_size = 0.2    # 测试集占比
bound = int(number_data * (1 - test_size))    # 分割训练集测试集的点，原本是浮点型，应转换成整数型

X_train = X[random_numbers[:bound], :]    # 行0,1,2,...,118,119，全部列
X_test = X[random_numbers[bound:], :]    # 行120,121,...,148,149，全部列
y_train = y[random_numbers[:bound]]    # 行0,1,2,...,118,119
y_test = y[random_numbers[bound:]]    # 行120,121,...,148,149

# 打印部分数据
print(f"打乱的序列为（前10个）：{random_numbers[:10]}")
print(f"测试集与训练集的分界点：{bound}")
print(f"特征矩阵训练集的形状：{X_train.shape}\n标签矩阵训练集的形状：{y_train.shape}")
print(f"特征矩阵测试集的形状：{X_test.shape}\n标签矩阵测试集的形状：{y_test.shape}")
```

注意：

此处，$M=3$，$N=120$。前者为变量`number_feature`，后者为变量`bound`

```python
# 初始化模型参数
learn_rate = 0.05    # 学习率
w = np.zeros((number_feature, 1))    # 权重，3*1
b = 0    # 截距
ones = np.ones((bound, 1))    # 全1列向量
loss_history = []    # 记录损失函数的变化，用于绘图
w_history = w.copy().reshape(1, -1)    # 记录权重的变化，用于求出变化量。若不加copy()，则后续其值会随着w一起变
b_history = [b]    # 记录截距的变化，用于求出变化量
max_delta = 1e-7    # max(|权重|,|截距|)相邻变化的临界值，低于此值即满足训练要求
upper_time = 2000    # 训练次数上限

# 打印部分数据
print(f"初始权重：\n{w}")
print(f"全1列向量（前5行）：\n{ones[:5]}\n形状：{ones.shape}")
print(b_history)
```

```python
# 训练模型
for epoch in np.arange(1, upper_time):    # 训练次数，1,2,..,1999,2000
    y_hat = X_train @ w + b    # 当前次数的估计值
    loss = np.mean((y_hat - y_train) ** 2)    # 当前训练次数的损失函数的值
    loss_history.append(loss)    # 保存当前损失函数的值
    
    w -= learn_rate * 2 / number_data * (X_train.T) @ (X_train @ w + b * ones - y_train)    # 迭代权重
    b -= learn_rate * 2 / number_data * (ones.T) @ (X_train @ w + b * ones - y_train)    # 迭代截距
    
    w_history = np.concatenate((w_history, w.reshape(1, -1)))    # 将新得出的权重w合并进去
    b_history.append(b[0][0])    #保存当前截距的值
    """
    得出来的权重w为列向量，应将其转为行向量，便于拼接
    得出来的截距b是二维数组[[0]]，需将其变成数
    """

    delta_w = np.abs((w_history[epoch, :]-w_history[epoch-1, :]))   # 权重变化量
    delta_b = np.abs(b_history[epoch]-b_history[epoch-1])    # 截距变化量
    delta_w_max = np.max(delta_w)    # 先求出截距变化的最大值

    delta = np.max([delta_w_max, delta_b])    # 最大的变化量
    
    # 每50次打印一次结果
    if epoch % 50 == 0:
        print(f"第{epoch}次训练：\n权重w=\n{w}\n截距b={b}")

    # 停止训练条件之一：权重,截距变化量达到要求
    if delta < max_delta:
        print(f"\n训练达到要求，训练次数：{epoch}")
        break

    # 停止训练条件之二：训练次数达到上限
    if epoch == upper_time:
        print(f"训练次数达{upper_time}，已结束训练！")

# 展示结果
print(f"权重：{w[:, 0]}\n截距：{b[0][0]:.2f}")
print(f"最终方程：y = {w[0, 0]:.2f}x1 + {w[1, 0]:.2f}x2 + {w[2, 0]:.2f}x3 + {b[0][0]:.2f}")
```

```python
# 测试模型
y_pred = X_test @ w + b    # 预测值

# 计算均方误差
test_mes = np.mean((y_pred - y_test) ** 2)
print(f"测试集的均方误差为：{test_mes}")
```
