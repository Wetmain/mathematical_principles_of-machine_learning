## Lasso回归-多元线性回归
在最小二乘法的基础上，添加$L_1$正则化项。Lasso回归与岭回归一样，不会影响偏置$b$

有$N$个样本且每个样本有$M$个特征，损失函数为
$$
J\left(\mathbf{w}^{\mathrm{T}}\right)=\frac{1}{N}\sum_{i=1}^{N}\left(\hat{y}_{i}-y_{i}\right)^2+\lambda\sum_{j=1}^{M}\left|w_{j}\right|
$$
其中，$\lambda$为正则化参数，不是矩阵特征值的符号；右端第一项为预测误差项，第二项为正则化项，
$$
\hat{y}_{i}=\sum_{j=1}^{M}w_{j}x_{ij}+b=\begin{pmatrix}x_{i1}, & \cdots, & x_{ij}, & \cdots, & x_{iM}\end{pmatrix}\begin{pmatrix}w_1\\ \cdots\\ w_{j}\\ \cdots\\ w_{M}\end{pmatrix}+b=\alpha_{i}\mathbf{w}+b
$$

---
现在必须要引入一些新概念，在下面不加证明地给出一些定义和结论

**凸函数**定义：
> 设函数：$f:\mathrm{R}^{n}\to\mathrm{R}$。若对于任意两个向量$\mathbf{x}_1,\mathbf{x}_2\in\mathrm{R}^{n}$，及任意数$\theta\in\left[0,1\right]$，都有
> $$
f\left(\theta\mathbf{x}_1+\left(1-\theta\right)\mathbf{x}_2\right)\le\theta f\left(\mathbf{x}_1\right)+\left(1-\theta\right)f\left(\mathbf{x}_2\right)
> $$
> 则称$f$为凸函数

注意：在几何上看，凸函数图像上任意两点的连线，不会在凸函数图像的下方

**Hessian矩阵**定义：
> 多元函数$f\left(\mathbf{x}\right)=f\left(x_1,x_2,\cdots,x_{n}\right)$的Hession矩阵为
> $$
\mathbf{H}\left(\mathbf{x}\right)=\begin{pmatrix}\frac{\partial^2f}{\partial x_1^2} & \cdots & \frac{\partial^2f}{\partial x_1\partial x_{n}}\\ \cdots & \cdots & \cdots\\ \frac{\partial^2f}{\partial x_{n}\partial x_1} & \cdots & \frac{\partial^2f}{\partial x_{n}^2}\end{pmatrix}
$$
**用Hessian矩阵判断凸函数**：
> 若函数$f\left(\mathbf{x}\right)$在凸区域上二阶连续可微，且其Hessian矩阵$\mathbf{H}\left(\mathbf{x}\right)\ge\mathbf{O}$，即矩阵半正定，则该函数为凸函数

**结论**：
> 1. 凸函数的局部极小值就是全局极小值
> 2. 两个凸函数之和仍为凸函数（这个很好理解，对应的Hessian矩阵都是半正定的，微分是线性运算，函数相加也就是矩阵相加。矩阵相加后仍为半正定的）

**次梯度**定义：
> 对于一个凸函数$f\left(\mathbf{x}\right)$，若存在一个数$g$，使定义域内任意的参数$\mathbf{z}$，满足
> $$
f\left.\left(\mathbf{z}\right.\right)\ge f\left(\mathbf{x}\right)+g\left(\mathbf{z}-\mathbf{x}\right)
> $$
> 则称该数$g$为凸函数$f$在$x$处的一个次梯度

注意：$g\left(z-x\right)$是$z=x$处在凸函数$f$下方的一条直线，该直线允许与凸函数重合

**次微分**定义：
> 对于凸函数$f\left(\mathbf{z}\right)$，其在一个点处的所有次梯度构成的集合，用$\partial f\left.\left(\mathbf{x}\right.\right)$表示

**例**：求凸函数$f\left.\left(x\right.\right)=\left|x\right|$的次微分

当$x>0$时，根据次梯度不等式定义，有$\left|z\right|\ge x+g\cdot\left(z-x\right)$

此时及$x$其他取值的情况，应将$x$固定，讨论任意的$z$对$g$的取值的影响

1. 当$z>x(>0)$时，得$z-x\ge g\cdot\left(z-x\right)\Rightarrow g\le1$

2. 当$0<z<x$时，得$z-x\ge g\cdot\left(z-x\right)\Rightarrow g\ge1$

先对这两种情况的$g$的取值区间取交集，可知$g=1$

3. 当$z=x$时，将$g=1$代入，得$0\ge g\cdot0$，这个不等式对于$g=1$时成立的

4. 当$z\le0$时，将$g=1$代入，得$-z\ge x+z-x\Rightarrow-z\ge z$，这也是成立的

可得$g=1$，当$x>0$时；

当$x<0$时，有$\left|z\right|\ge-x+g\cdot\left(z-x\right)$

1. 当$z<x(<0)$时，得$-z+x\ge g\cdot\left(z-x\right)\Rightarrow g\ge-1$

2. 当$x<z<0$时，得$-z+x\ge g\cdot\left(z-x\right)\Rightarrow g\le-1$

先对这两种情况的$g$的取值区间取交集，得$g=-1$

3. 当$z$为其他取值时，$g=-1$也符合不等式

可得$g=-1$，当$x<0$时；

当$x=0$时，次梯度不等式中，直线的斜率可以取$\left[-1,1\right]$，也就是$g\in\left[-1,1\right]$

现在证明，将$x=0$代入次梯度不等式，得
$$
\begin{align*} & f\left.\left(z\right.\right)\ge f\left(0\right)+g\left(z-0\right)\\  & \Rightarrow\left|z\right|\ge gz\end{align*}
$$
1. 当$z>0$时，有$z\ge gz\Rightarrow g\le1$

2. 当$z<0$时，有$-z\ge gz\Rightarrow g\ge-1$

3. 当$z=0$时，任意$g$都使不等式成立

次梯度不等式需要对任意$z$都成立，需将上述各情况所得的$g$的取值区间取交集

可得$g\in\left[-1,1\right]$

则凸函数$f\left.\left(x\right.\right)=\left|x\right|$的次微分为
$$
\partial f\left(x\right)=\begin{cases}1 & ,x>0\\ \left[-1,1\right] & ,x=0\\ -1 & ,x<0\end{cases}
$$
**结论**：
> 1. 对于定义在定义在整个$\mathrm{R}^{n}$上且取值有限的凸函数$f(x)$，其无约束优化问题满足
> $$
x^{\ast}是全局最优解\Leftrightarrow0\in\partial f\left(x^{\ast}\right)
$$

现在给出一些矩阵微积分的公式
1. **标量对向量的导数**：也就是梯度

设标量函数$f\left(\mathbf{x}\right)=f\begin{pmatrix}x_1, & \cdots, & x_{i}, & \cdots, & x_{n}\end{pmatrix}$

其中，
$$
\mathbf{x}=\begin{pmatrix}x_1\\ \cdots\\ x_{i}\\ \cdots\\ _{n}\end{pmatrix}
$$
有
> $$
\nabla_{\mathbf{x}}f\left(\mathbf{x}\right)=\begin{pmatrix}\frac{\partial f}{\partial x_1}\\ \cdots\\ \frac{\partial f}{\partial x_{i}}\\ \cdots\\ \frac{\partial f}{\partial x_{n}}\end{pmatrix}
$$
2. **向量对向量的导数**：

设由标量函数$f_{i}\left(\mathbf{x}\right)$组成的向量函数
$$
\mathbf{f}\left(\mathbf{x}\right)=\begin{pmatrix}f_1\left(\mathbf{x}\right)\\ \cdots\\ f_{i}\left(\mathbf{x}\right)\\ \cdots\\ f_{n}\left(\mathbf{x}\right)\end{pmatrix}
$$
有
> $$
\frac{\partial\mathbf{f}}{\partial\mathbf{x}}=\begin{pmatrix}\frac{\partial f_1}{\partial x_1} & \cdots & \frac{\partial f_1}{\partial x_{n}}\\ \cdots & \cdots & \cdots\\ \frac{\partial f_{m}}{\partial x_1} & \cdots & \frac{\partial f_{m}}{\partial x_{n}}\end{pmatrix}=\begin{pmatrix}\left(\nabla_{\mathbf{x}}f_1\right)^{\mathrm{T}}\\ \cdots\\ \left(\nabla_{\mathbf{x}}f_{i}\right)^{\mathrm{T}}\\ \cdots\\ \left(\nabla_{\mathbf{x}}f_{m}\right)^{\mathrm{T}}\end{pmatrix}
$$

---
现在给出一些机器学习中常用的向量导数公式
1. **线性函数**：$\mathbf{a}$常列向量，有
> $$
\nabla_{\mathbf{x}}\left(\mathbf{a}^{\mathrm{T}}\mathbf{x}\right)=\mathbf{a}
$$
2. **向量线性变换**：其中常矩阵$\mathbf{A}$称为Jacobian矩阵，
> $$
\frac{\partial}{\partial\mathbf{x}}\left(\mathbf{Ax}\right)=\mathbf{A}
$$
3. **向量的平方范数**：其中$\mathbf{x}^{\mathrm{T}}\mathbf{x}=x_1^2+x_{i}^2+\cdots+x_{n}^2$
> $$
\frac{\partial}{\partial\mathbf{x}}\left(\mathbf{x}^{\mathrm{T}}\mathbf{x}\right)=2\mathbf{x}
$$
4. **二次型**：其中$\mathbf{A}$为$n$阶常矩阵
> $$
\frac{\partial}{\partial\mathbf{x}}\left(\mathbf{x}^{\mathrm{T}}\mathbf{Ax}\right)=\left(\mathbf{A}+\mathbf{A}^{\mathrm{T}}\right)\mathbf{x}
$$
5. **复合二次函数**：$\left(\mathbf{Ax}-\mathbf{b}\right)^{\mathrm{T}}\left(\mathbf{Ax}-\mathbf{b}\right)=\|\mathbf{Ax}-\mathbf{b}\|_2^2$，其中$\mathbf{A},\mathbf{x}$分别为常矩阵,常向量
> $$
\nabla_{\mathbf{x}}\left[\left(\mathbf{Ax}-\mathbf{b}\right)^{\mathrm{T}}\left(\mathbf{Ax}-\mathbf{b}\right)\right]=2\mathbf{A}^{\mathrm{T}}\left(\mathbf{Ax}-\mathbf{b}\right)
$$

现在推导损失函数预测误差的梯度。这部分在最后推导权重更新公式时不会涉及到

接下来会用到的向量运算公式：
$$
\begin{align*} & \sum_{i=1}^{N}x_{i}^2=\mathbf{x}^{\mathrm{T}}\mathbf{x},\mathbf{x}=\begin{pmatrix}x_1\\ \cdots\\ x_{i}\\ \cdots\\ x_{N}\end{pmatrix}\\  & \mathbf{x}^{\mathrm{T}}\mathbf{y}=\mathbf{y}^{\mathrm{T}}\mathbf{x},\mathbf{x}=\begin{pmatrix}x_1\\ \cdots\\ x_{i}\\ \cdots\\ x_{N}\end{pmatrix},\mathbf{y}=\begin{pmatrix}y_1\\ \cdots\\ y_{i}\\ \cdots\\ y_{N}\end{pmatrix}\end{align*}
$$
对于右侧第一项求和式，有
$$
\begin{align*} & \sum_{i=1}^{N}\left(\alpha_{i}\mathbf{w}+b-y_{i}\right)^2=\begin{pmatrix}\alpha_1\mathbf{w}+b-y_1\\ \cdots\\ \alpha_{i}\mathbf{w}+b-y_{i}\\ \cdots\\ \alpha_{N}\mathbf{w}+b-y_{N}\end{pmatrix}^{\mathrm{T}}\begin{pmatrix}\alpha_1\mathbf{w}+b-y_1\\ \cdots\\ \alpha_{i}\mathbf{w}+b-y_{i}\\ \cdots\\ \alpha_{N}\mathbf{w}+b-y_{N}\end{pmatrix}\\  & =\begin{pmatrix}\alpha_1\mathbf{w}+b-y_1\\ \cdots\\ \alpha_{i}\mathbf{w}+b-y_{i}\\ \cdots\\ \alpha_{N}\mathbf{w}+b-y_{N}\end{pmatrix}^{\mathrm{T}}\left[\begin{pmatrix}\alpha_1\\ \cdots\\ \alpha_{i}\\ \cdots\\ \alpha_{N}\end{pmatrix}\mathbf{w}+\begin{pmatrix}b\\ \cdots\\ b\\ \cdots\\ b\end{pmatrix}-\begin{pmatrix}y_1\\ \cdots\\ y_{i}\\ \cdots\\ y_{N}\end{pmatrix}\right]=\begin{pmatrix}\alpha_1\mathbf{w}+b-y_1\\ \cdots\\ \alpha_{i}\mathbf{w}+b-y_{i}\\ \cdots\\ \alpha_{N}\mathbf{w}+b-y_{N}\end{pmatrix}^{\mathrm{T}}\left(\mathbf{Xw}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)\\  & =\left(\mathbf{Xw}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)^{\mathrm{T}}\left(\mathbf{Xw}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)=\left(\mathbf{w}^{\mathrm{T}}\mathbf{X}^{\mathrm{T}}+b\left(\mathbf{1}\right)_{N}^{\mathrm{T}}-\mathbf{y}^{\mathrm{T}}\right)\left(\mathbf{Xw}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)\\  & =\mathbf{w}^{\mathrm{T}}\mathbf{X}^{\mathrm{T}}\mathbf{Xw}+\mathbf{w}^{\mathrm{T}}\mathbf{X}^{\mathrm{T}}\left(b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)+b\left(\mathbf{1}\right)_{N}^{\mathrm{T}}\mathbf{Xw}+b\left(\mathbf{1}\right)_{N}^{\mathrm{T}}\left(b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)-\mathbf{y}^{\mathrm{T}}\mathbf{Xw}-\mathbf{y}^{\mathrm{T}}\left(b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)\\  & =\mathbf{w}^{\mathrm{T}}\mathbf{X}^{\mathrm{T}}\mathbf{Xw}+\mathbf{w}^{\mathrm{T}}\mathbf{X}^{\mathrm{T}}\left(b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)+\mathbf{w}^{\mathrm{T}}\mathbf{X}^{\mathrm{T}}b\left(\mathbf{1}\right)_{N}+\mathbf{w}^{\mathrm{T}}\mathbf{X}^{\mathrm{T}}\left(-\mathbf{y}\right)+\left(b\left(\mathbf{1}\right)_{N}^{\mathrm{T}}-\mathbf{y}^{\mathrm{T}}\right)\left(b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)\\  & =\mathbf{w}^{\mathrm{T}}\mathbf{X}^{\mathrm{T}}\mathbf{Xw}+2\left(b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)^{\mathrm{T}}\mathbf{Xw}+C\end{align*}
$$
其中，$C$是与$\mathbf{w}$无关的常量，及
$$
\mathbf{X}=\begin{pmatrix}\alpha_1\\ \cdots\\ \alpha_{i}\\ \cdots\\ \alpha_{N}\end{pmatrix}=\begin{pmatrix}x_{11} & \cdots & x_{1j} & \cdots & x_{1M}\\ \cdots & \cdots & \cdots & \cdots & \cdots\\ x_{i1} & \cdots & x_{ij} & \cdots & x_{iM}\\ \cdots & \cdots & \cdots & \cdots & \cdots\\ x_{N1} & \cdots & x_{Nj} & \cdots & x_{NM}\end{pmatrix}
$$

再根据上面给出的标量函数的导数公式，得
$$
\nabla_{\mathbf{w}}\left[\sum_{i=1}^{N}\left(\alpha_{i}\mathbf{w}+b-y_{i}\right)^2\right]=2\mathbf{X}^{\mathrm{T}}\mathbf{Xw}+2\left(b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)^{\mathrm{T}}\mathbf{X}
$$

---
对于损失函数右侧第二项求和式，在$w_{j}=0$处不可导，因此需要用到次微分的方法

而且，Lasso回归不能像岭回归那样，直接得出全部权重$\mathbf{w}$的表达式，原因如下

岭回归的$L_2$正则化项求导后为$2\lambda\mathbf{I}_{M}\mathbf{w}$，可以与预测误差项合并，得到类似于$\mathbf{Aw}=\mathbf{b}$的形式

而Lasso回归的$L_1$正则化项，求导后的形式为$\mathrm{sign}(\mathbf{w})$，先认为权重都不为零，这无法与预测误差项合并，故不能直接求解

其中，$\mathrm{sign}\left(x\right)=\begin{cases}1 & ,x>0\\ -1 & ,x<0\end{cases}$为符号函数，不考虑$x=0$

要求得权重的更新公式，可选择**坐标下降法**：对于多个权重$w_1,w_2,\cdots ,w_{M}$，每次只更新一个参数值，其余参数保持不变。如指定初始的$w_1,w_2,\cdots ,w_{M}$后，先更新$w_1$，再用新的$w_1$和之前的$,w_3,\cdots ,w_{M}$更新$w_2$，一直更新完所有的，权重。如此进行下去，知道满足收敛条件

因为权重是一个个地更新，所以可以先推导一元线性回归公式的**软阈值函数**，这可以用到多元线性回归中

一元线性回归的Lasso损失函数为
> $$
\begin{align*} & J\left(w\right)=\frac{1}{N}\sum_{i=1}^{N}\left(wx_{i}+b-y_{i}\right)^2+\lambda\left|w\right|\\  & =\left(\frac{1}{N}\sum_{i=1}^{N}x_{i}^2\right)w^2+2\left[\frac{1}{N}\sum_{i=1}^{N}\left(b-y_{i}\right)\right]w+\frac{1}{N}\sum_{i=1}^{N}\left(b-y_{i}\right)^2+\lambda\left|w\right|\\  & =\frac12aw^2-cw+\lambda\left|w\right|+C\end{align*}
$$
其中，认为所有$x_{i}$不同时为零
$$
a=\frac{2}{N}\sum_{i=1}^{N}x_{i}^2>0,c=-\frac{2}{N}\sum_{i=1}^{N}\left(b-y_{i}\right),C=\frac{1}{N}\sum_{i=1}^{N}\left(b-y_{i}\right)^2
$$
对损失函数是个凸函数，且在$x=0$处不可微，应该求次微分，得
$$
\partial J\left(w\right)=aw-c+\lambda\partial\left|w\right|
$$
使用上面的结论，有最优解等价于“$0$”属于次微分集合，即
$$
0\in\partial J\left(w\right)=aw-c+\lambda\partial\left|w\right|
$$
1. 当$w>0$时，$\partial\left|w\right|={1}$，即$\partial J\left(w\right)=aw-c+\lambda\left\{1\right\}$

次微分只有一个元素，上面的$0\in\partial J\left(w\right)$就变成了$0=\partial J\left(w\right)$

得$0=aw-c+\lambda\Rightarrow w=\frac{c-\lambda}{a}$

满足$0<w=\frac{c-\lambda}{a}$，要求$c>\lambda$

2. 当$w<0$时，$\partial\left|w\right|={-1}$，即$\partial J\left(w\right)=aw-c+\lambda\left\{-1\right\}$

同理得$0=aw-c-\lambda\Rightarrow w=\frac{c+\lambda}{a}$

满足$0>w=\frac{c+\lambda}{a}$，要求$c<-\lambda$

3. 当$w=0$时，$\partial\left|w\right|=\left[-1,1\right]$，即$\partial J\left(w\right)=-c+\lambda\left[-1,1\right]$，注意$w=0$也要代入

得$0\in-c-\lambda\left[-1,1\right]\Rightarrow c\in\left[-\lambda,\lambda\right]\Rightarrow\left|c\right|\le\lambda$

由上述3种情况，故
> $$
w=\begin{cases}\frac{c-\lambda}{a} & ,c>\lambda\\ 0 & ,\left|c\right|\le\lambda\\ \frac{c+\lambda}{a} & ,c<-\lambda\end{cases}=\frac{1}{a}\begin{cases}c-\lambda & ,c>\lambda\\ 0 & ,\left|c\right|\le\lambda\\ c+\lambda & ,c<-\lambda\end{cases}=\frac{S\left(c,\lambda\right)}{a}
$$
其中，$S\left(c,\lambda\right)=\mathrm{sign}\left(c\right)\max\left\{\left|c\right|-\lambda,0\right\}$称为软阈值函数


现在将损失函数换个写法。对于第一项求和式有
$$
\begin{align*} & \sum_{i=1}^{N}\left(\alpha_{i}\mathbf{w}+b-y_{i}\right)^2=\begin{pmatrix}\alpha_1\mathbf{w}+b-y_1\\ \cdots\\ \alpha_{i}\mathbf{w}+b-y_{i}\\ \cdots\\ \alpha_{N}\mathbf{w}+b-y_{N}\end{pmatrix}^{\mathrm{T}}\begin{pmatrix}\alpha_1\mathbf{w}+b-y_1\\ \cdots\\ \alpha_{i}\mathbf{w}+b-y_{i}\\ \cdots\\ \alpha_{N}\mathbf{w}+b-y_{N}\end{pmatrix}\\  & =\|\begin{pmatrix}\sum_{j=1}^{M}w_{j}x_{1j}+b-y_1\\ \cdots\\ \sum_{j=1}^{M}w_{j}x_{ij}+b-y_{i}\\ \cdots\\ \sum_{j=1}^{M}w_{j}x_{Nj}+b-y_{N}\end{pmatrix}\|_2^2=\|\begin{pmatrix}w_1x_{11}+\cdots+w_{M}x_{1M}\\ \cdots\\ w_1x_{11}+\cdots+w_{M}x_{1M}\\ \cdots\\ w_1x_{11}+\cdots+w_{M}x_{1M}\end{pmatrix}+\begin{pmatrix}b\\ \cdots\\ b\\ \cdots\\ b\end{pmatrix}-\begin{pmatrix}y_1\\ \cdots\\ y_{i}\\ \cdots\\ y_{N}\end{pmatrix}\|_2^2\\  & =\|\begin{pmatrix}w_1x_{11}\\ \cdots\\ w_1x_{i1}\\ \cdots\\ w_1x_{N1}\end{pmatrix}+\cdots+\begin{pmatrix}w_{M}x_{1M}\\ \cdots\\ w_{M}x_{iM}\\ \cdots\\ w_{M}x_{NM}\end{pmatrix}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\|_2^2=\|w_1\begin{pmatrix}x_{11}\\ \cdots\\ x_{i1}\\ \cdots\\ x_{N1}\end{pmatrix}+\cdots+w_{M}\begin{pmatrix}x_{1M}\\ \cdots\\ x_{iM}\\ \cdots\\ x_{NM}\end{pmatrix}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\|_2^2\\  & =\|w_1\mathbf{x}_1+\cdots+w_{M}\mathbf{x}_{M}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\|_2^2=\|\sum_{j=1}^{M}w_{j}\mathbf{x}_{j}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\|_2^2\end{align*}
$$
其中，$\|\cdot\|_2^2$表示2-范数的平方，再次只是借用表示方法，及
$$
\mathbf{x}_{j}=\begin{pmatrix}x_{1j}\\ \cdots\\ x_{ij}\\ \cdots\\ x_{Nj}\end{pmatrix}
$$

定义**部分残差**：$\mathbf{r}_{j}$表示去掉第$j$个特征的贡献后所得的残差，表达式为
> $$
\mathbf{r}_{j}=\sum_{k\ne j}^{M}w_{k}\mathbf{x}_{k}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}
$$
显然，部分残差$\mathbf{r}_{j} $与
$w_{j}$无关

代入上面的式子，得
$$
\begin{align*} & \sum_{i=1}^{N}\left(\alpha_{i}\mathbf{w}+b-y_{i}\right)^2=\|\sum_{j=1}^{M}w_{j}\mathbf{x}_{j}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\|_2^2=\|\sum_{k\ne j}^{M}w_{k}\mathbf{x}_{k}+w_{j}\mathbf{x}_{j}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\|_2^2=\|\mathbf{r}_{j}+w_{j}\mathbf{x}_{j}\|_2^2\end{align*}
$$
于是，损失函数可以表示为
$$
J\left(\mathbf{w}^{\mathrm{T}}\right)=\frac{1}{N}\sum_{i=1}^{N}\left(\alpha_{i}\mathbf{w}+b-y_{i}\right)^2+\lambda\sum_{j=1}^{M}\left|w_{j}\right|=\frac{1}{N}\|\mathbf{r}_{j}+w_{j}\mathbf{x}_{j}\|_2^2+\lambda\left|w_{j}\right|+\lambda\sum_{k\ne j}^{M}\left|w_{k}\right|
$$
现在更新$w_{j}$，此时损失函数仅与$w_{j}$有关，得
> $$
\begin{align*} & J\left(w_{j}\right)=\frac{1}{N}\|\mathbf{r}_{j}+w_{j}\mathbf{x}_{j}\|_2^2+\lambda\left|w_{j}\right|+\lambda\sum_{k\ne j}^{M}\left|w_{k}\right|\\  & =\frac{1}{N}\left(\mathbf{r}_{j}+w_{j}\mathbf{x}_{j}\right)^{\mathrm{T}}\left(\mathbf{r}_{j}+w_{j}\mathbf{x}_{j}\right)+\lambda\left|w_{j}\right|+\lambda\sum_{k\ne j}^{M}\left|w_{k}\right|\\  & =\frac{1}{N}\left(\mathbf{r}_{j}^{\mathrm{T}}+w_{j}\mathbf{x}_{j}^{\mathrm{T}}\right)\left(\mathbf{r}_{j}+w_{j}\mathbf{x}_{j}\right)+\lambda\left|w_{j}\right|+\lambda\sum_{k\ne j}^{M}\left|w_{k}\right|\\  & =\frac{1}{N}\left(\mathbf{r}_{j}^{\mathrm{T}}\mathbf{r}_{j}+2w_{j}\mathbf{r}_{j}^{\mathrm{T}}\mathbf{x}_{j}+w_{j}^2\mathbf{x}_{j}^{\mathrm{T}}\mathbf{x}_{j}\right)+\lambda\left|w_{j}\right|+\lambda\sum_{k\ne j}^{M}\left|w_{k}\right|\end{align*}
$$
与上面一元线性回归的Lasso损失函数，相比较，得
$$
\begin{align*} & a=\frac{2}{N}\mathbf{x}_{j}^{\mathrm{T}}\mathbf{x}_{j}\\  & c=-\frac{2}{N}\mathbf{r}_{j}^{\mathrm{T}}\mathbf{x}_{j}\\  & C=\frac{2}{N}\mathbf{r}_{j}^{\mathrm{T}}\mathbf{r}_{j}+\lambda\sum_{k\ne j}^{M}\left|w_{k}\right|\end{align*}
$$
