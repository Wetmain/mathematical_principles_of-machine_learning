## 最小二乘法-多元线性回归
有$N$个样本且每个样本有$M$个特征，选择的损失函数为
$$
J\left(\mathbf{w}^{\mathrm{T}},b\right)=\frac{1}{N}\sum_{i=1}^{N}\left(\hat{y}_{i}-y_{i}\right)^2=\frac{1}{N}\sum_{i=1}^{N}\left(w_1x_{i1}+\cdots+w_{j}x_{ij}+\cdots+w_{M}x_{iM}+b-y_{i}\right)^2
$$
令$x_{i(M+1)}=1,w_{M+1}=b$，可得
$$
J\left(\mathbf{w}^{\mathrm{T}},b\right)=\frac{1}{N}\sum_{i=1}^{N}\left(\hat{y}_{i}-y_{i}\right)^2=\frac{1}{N}\sum_{i=1}^{N}\left(w_1x_{i1}+\cdots+w_{j}x_{ij}+\cdots+w_{M}x_{iM}+w_{M+1}x_{i(M+1)}-y_{i}\right)^2
$$
其中，有
$$
\begin{align*} & w_1x_{i1}+\cdots+w_{j}x_{ij}+\cdots+w_{M+1}x_{\left.i\left(M+1\right.\right)}-y_{i}\\  & =\begin{pmatrix}x_{i1}, & \cdots, & x_{ij}, & \cdots, & x_{i\left(M+1\right)}\end{pmatrix}\begin{pmatrix}w_1\\ \cdots\\ w_{j}\\ \cdots\\ w_{M+1}\end{pmatrix}-y_{i}\\  & =\alpha_{i}\mathbf{W}-y_{i}\end{align*}
$$

---
现在推导权重,截距的表达式，也就是求$\mathbf{W}$

最小二乘法是令偏导等于零，求偏导，得
$$
\frac{\partial}{\partial w_{j}}J\left(\mathbf{w}^{\mathrm{T}},b\right)=\frac{2}{N}\sum_{i=1}^{N}\left(w_1x_{i1}+\cdots+w_{j}x_{ij}+\cdots+w_{M+1}x_{i(M+1)}-y_{i}\right)x_{ij}=\frac{2}{N}\sum_{i=1}^{N}\left(\alpha_{i}\mathbf{W}-y_{i}\right)x_{ij}
$$
对于求和式，有
$$
\begin{align*} & \sum_{i=1}^{N}\left(\alpha_{i}\mathbf{W}-y_{i}\right)x_{ij}=\begin{pmatrix}x_{1j}, & \cdots, & x_{ij}, & \cdots, & x_{Nj}\end{pmatrix}\begin{pmatrix}\alpha_1\mathbf{W}-y_1\\ \cdots\\ \alpha_{i}\mathbf{W}-y_{i}\\ \cdots\\ \alpha_{N}\mathbf{W}-y_{N}\end{pmatrix}\\  & =\begin{pmatrix}x_{1j}, & \cdots, & x_{ij}, & \cdots, & x_{Nj}\end{pmatrix}\begin{pmatrix}\alpha_1\mathbf{W}-y_1\\ \cdots\\ \alpha_{i}\mathbf{W}-y_{i}\\ \cdots\\ \alpha_{N}\mathbf{W}-y_{N}\end{pmatrix}=\beta_{j}\left[\begin{pmatrix}\alpha_1\mathbf{W}\\ \cdots\\ \alpha_{i}\mathbf{W}\\ \cdots\\ \alpha_{N}\mathbf{W}\end{pmatrix}-\begin{pmatrix}y_1\\ \cdots\\ y_{i}\\ \cdots\\ y_{N}\end{pmatrix}\right]\\  & =\beta_{j}\left[\begin{pmatrix}\alpha_1\\ \cdots\\ \alpha_{i}\\ \cdots\\ \alpha_{N}\end{pmatrix}\mathbf{W}-\begin{pmatrix}y_1\\ \cdots\\ y_{i}\\ \cdots\\ y_{N}\end{pmatrix}\right]=\beta_{j}\left(\mathbf{XW}-\mathbf{y}\right)\end{align*}
$$
其中，$j=1,2,\cdots ,M+1$，且
$$
\beta_{j}=\begin{pmatrix}x_{1j}, & \cdots, & x_{ij}, & \cdots, & x_{Nj}\end{pmatrix},\mathbf{X}=\begin{pmatrix}\alpha_1\\ \cdots\\ \alpha_{i}\\ \cdots\\ \alpha_{N}\end{pmatrix}=\begin{pmatrix}x_{11} & \cdots & x_{1j} & \cdots & x_{1\left(M+1\right)}\\ \cdots & \cdots & \cdots & \cdots & \cdots\\ x_{i1} & \cdots & x_{ij} & \cdots & x_{i\left(M+1\right)}\\ \cdots & \cdots & \cdots & \cdots & \cdots\\ x_{N1} & \cdots & x_{Nj} & \cdots & x_{N\left(M+1\right)}\end{pmatrix}
$$
于是偏导为
$$
\frac{\partial}{\partial w_{j}}J\left(\mathbf{w}^{\mathrm{T}},b\right)=\frac{2}{N}\sum_{i=1}^{N}\left(\alpha_{i}\mathbf{W}-y_{i}\right)x_{ij}=\frac{2}{N}\beta_{j}\left(\mathbf{XW}-\mathbf{y}\right)
$$
令各个偏导等于零，得
$$
\frac{2}{N}\beta_{j}\left(\mathbf{XW}-\mathbf{y}\right)=0,j=1,2,\cdots ,M+1
$$
拼成矩阵，有
$$
\begin{pmatrix}0\\ \cdots\\ 0\end{pmatrix}=\mathbf{O}=\frac{\partial}{\partial\mathbf{W}}J\left(\mathbf{w}^{\mathrm{T}},b\right)=\frac{2}{N}\begin{pmatrix}\beta_1\left(\mathbf{XW}-\mathbf{y}\right)\\ \cdots\\ \beta_{j}\left(\mathbf{XW}-\mathbf{y}\right)\\ \cdots\\ \beta_{M+1}\left(\mathbf{XW}-\mathbf{y}\right)\end{pmatrix}=\frac{2}{N}\begin{pmatrix}\beta_1\\ \cdots\\ \beta_{j}\\ \cdots\\ \beta_{M+1}\end{pmatrix}\left(\mathbf{XW}-\mathbf{y}\right)
$$
其中，有
$$
\begin{pmatrix}\beta_1\\ \cdots\\ \beta_{j}\\ \cdots\\ \beta_{M+1}\end{pmatrix}=\begin{pmatrix}x_{11} & \cdots & x_{i1} & \cdots & x_{N1}\\ \cdots & \cdots & \cdots & \cdots & \cdots\\ x_{1j} & \cdots & x_{ij} & \cdots & x_{Nj}\\ \cdots & \cdots & \cdots & \cdots & \cdots\\ x_{1\left(M+1\right)} & \cdots & x_{i\left(M+1\right)} & \cdots & x_{N\left(M+1\right)}\end{pmatrix}=\mathbf{X}^{\mathrm{T}}
$$
得
$$
\frac{2}{N}\mathbf{X}^{\mathrm{T}}\left(\mathbf{XW}-\mathbf{y}\right)=\mathbf{O}
$$
即
$$
\mathbf{X}^{\mathrm{T}}\mathbf{XW}=\mathbf{X}^{\mathrm{T}}\mathbf{y}
$$
于是解得权重,截距为
$$
\mathbf{W}=\left(\mathbf{X}^{\mathrm{T}}\mathbf{X}\right)^{-1}\mathbf{X}^{\mathrm{T}}\mathbf{y}
$$

事实上，$\mathbf{X}^{\mathrm{T}}\mathbf{X}$并不一定可逆，这是个实对称矩阵，且是半正定的

现在证明其为半正定的，构造二次型$f\left(\mathbf{z}\right)=\mathbf{z}^{\mathrm{T}}\left(\mathbf{X}^{\mathrm{T}}\mathbf{X}\right)\mathbf{z}$，
其中，
$$
\mathbf{z}=\begin{pmatrix}z_1\\ \cdots\\ z_{M+1}\end{pmatrix}
$$
要证明$\mathbf{X}^{\mathrm{T}}\mathbf{X}$为半正定矩阵，就是要证明$f\left(\mathbf{z}\right)$为半正定二次型，

即对于任意非零的$\mathbf{z}$，都有$f\left(\mathbf{z}\right)\ge 0$

根据矩阵运算，得
$$
\mathbf{z}^{\mathrm{T}}\left(\mathbf{X}^{\mathrm{T}}\mathbf{X}\right)\mathbf{z}=\mathbf{z}^{\mathrm{T}}\mathbf{X}^{\mathrm{T}}\mathbf{Xz}=\left(\mathbf{Xz}\right)^{\mathrm{T}}\mathbf{Xz}=\left(\mathbf{Xz},\mathbf{Xz}\right)
$$
其中$\mathbf{z}$为任意非零的列向量

这就是向量$\mathbf{Xz}$模长的平方，显然非负

故对于任意的非零列向量$\mathbf{z}$，都有
$$
f\left(\mathbf{z}\right)\ge 0
$$
从而得出$\mathbf{X}^{\mathrm{T}}\mathbf{X}$是半正定矩阵
