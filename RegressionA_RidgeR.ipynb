## 岭回归-多元线性回归
在最小二乘法的损失函数的基础上，添加$L_2$正则化项。因为岭回归不会影响截距$b$，故损失函数不是截距$b$的函数

有$N$个样本且每个样本有$M$个特征，损失函数为
$$
J\left(\mathbf{w}^{T}\right)=\frac{1}{N}\sum_{i=1}^{N}\left(\hat{y}_{i}-y_{i}\right)^2+\lambda\sum_{j=1}^{M}w_{j}^2
$$
其中，$\lambda$为正则化参数，不是矩阵特征值的符号；右端第一项为预测误差，第二项为正则化项，
$$
\hat{y}_{i}=\sum_{j=1}^{M}w_{j}x_{ij}+b=\begin{pmatrix}x_{1j}, & \cdots, & x_{ij}, & \cdots, & x_{Nj}\end{pmatrix}\begin{pmatrix}w_1\\ \cdots\\ w_{j}\\ \cdots\\ w_{M}\end{pmatrix}+b=\alpha_{j}\mathbf{w}+b
$$

---
现在推导$\mathbf{w}$的表达式。对$w_j$求导，得
$$
\frac{\partial}{\partial w_{j}}J\left(\mathbf{w}^{T}\right)=\frac{2}{N}\sum_{i=1}^{N}\left(\alpha_{j}\mathbf{w}+b-y_{i}\right)x_{ij}+2\lambda w_{j}
$$
对于求和式，有
$$
\begin{align*} & \sum_{i=1}^{N}\left(\alpha_{j}\mathbf{w}+b-y_{i}\right)x_{ij}=\begin{pmatrix}x_{1j}, & \cdots, & x_{ij}, & \cdots, & x_{Nj}\end{pmatrix}\begin{pmatrix}\alpha_1\mathbf{w}+b-y_1\\ \cdots\\ \alpha_{j}\mathbf{w}+b-y_{j}\\ \cdots\\ \alpha_{M}\mathbf{w}+b-y_{M}\end{pmatrix}\\  & =\beta_{j}\left[\begin{pmatrix}\alpha_1\mathbf{w}\\ \cdots\\ \alpha_{j}\mathbf{w}\\ \cdots\\ \alpha_{M}\mathbf{w}\end{pmatrix}+\begin{pmatrix}b\\ \cdots\\ b\\ \cdots\\ b\end{pmatrix}-\begin{pmatrix}y_1\\ \cdots\\ y_{j}\\ \cdots\\ y_{M}\end{pmatrix}\right]=\beta_{j}\left[\begin{pmatrix}\alpha_1\\ \cdots\\ \alpha_{j}\\ \cdots\\ \alpha_{M}\end{pmatrix}\mathbf{w}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right]=\beta_{j}\left(\mathbf{Xw}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)\end{align*}
$$
其中，
$$
\mathbf{X}=\begin{pmatrix}x_{11} & \cdots & x_{1j} & \cdots & x_{1M}\\ \cdots & \cdots & \cdots & \cdots & \cdots\\ x_{i1} & \cdots & x_{ij} & \cdots & x_{iM}\\ \cdots & \cdots & \cdots & \cdots & \cdots\\ x_{N1} & \cdots & x_{Nj} & \cdots & x_{NM}\end{pmatrix}
$$
拼成矩阵，有
$$
\begin{align*} & \frac{\partial}{\partial\mathbf{w}^{T}}J\left(\mathbf{w}^{T}\right)=\begin{pmatrix}\frac{\partial}{\partial w_1}J\left(\mathbf{w}^{T}\right)\\ \cdots\\ \frac{\partial}{\partial w_{j}}J\left(\mathbf{w}^{T}\right)\\ \cdots\\ \frac{\partial}{\partial w_{M}}J\left(\mathbf{w}^{T}\right)\end{pmatrix}=\begin{pmatrix}\frac{2}{N}\beta_1\left(\mathbf{Xw}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)+2\lambda w_1\\ \cdots\\ \frac{2}{N}\beta_{j}\left(\mathbf{Xw}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)+2\lambda w_{j}\\ \cdots\\ \frac{2}{N}\beta_{M}\left(\mathbf{Xw}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)+2\lambda w_{M}\end{pmatrix}\\  & =\frac{2}{N}\begin{pmatrix}\beta_1\left(\mathbf{Xw}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)\\ \cdots\\ \beta_{j}\left(\mathbf{Xw}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)\\ \cdots\\ \beta_{M}\left(\mathbf{Xw}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)\end{pmatrix}+2\lambda\begin{pmatrix}w_1\\ \cdots\\ w_{j}\\ \cdots\\ w_{M}\end{pmatrix}=\frac{2}{N}\left[\begin{pmatrix}\beta_1\\ \cdots\\ \beta_{j}\\ \cdots\\ \beta_{M}\end{pmatrix}\left(\mathbf{Xw}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)\right]+2\lambda\mathbf{w}\end{align*}
$$
其中，
$$
\begin{align*} & \begin{pmatrix}\beta_1\\ \cdots\\ \beta_{j}\\ \cdots\\ \beta_{M}\end{pmatrix}=\begin{pmatrix}x_{11} & \cdots & x_{i1} & \cdots & x_{N1}\\ \cdots & \cdots & \cdots & \cdots & \cdots\\ x_{1j} & \cdots & x_{ij} & \cdots & x_{Nj}\\ \cdots & \cdots & \cdots & \cdots & \cdots\\ x_{1M} & \cdots & x_{iM} & \cdots & x_{NM}\end{pmatrix}=\mathbf{X}^{T}\\  & \mathbf{w}=\mathbf{I}_{M}\mathbf{w},\mathbf{I}_{M}=\begin{pmatrix}1 &  &  &  & \\  & \cdots &  &  & \\  &  & 1 &  & \\  &  &  & \cdots & \\  &  &  &  & 1\end{pmatrix}_{M}\end{align*}
$$
得
$$
\frac{\partial}{\partial\mathbf{w}^{T}}J\left(\mathbf{w}^{T}\right)=\frac{2}{N}\mathbf{X}^{T}\left(\mathbf{Xw}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)+2\lambda\mathbf{I}_{M}\mathbf{w}
$$
令$\frac{\partial}{\partial\mathbf{w}^{T}}J\left(\mathbf{w}^{T}\right)=\mathbf{0}$，得
$$
\begin{align*} & \frac{2}{N}\mathbf{X}^{T}\left(\mathbf{Xw}+b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)+2\lambda\mathbf{I}_{M}\mathbf{w}=\mathbf{0}\\  & \left(\mathbf{X}^{T}\mathbf{X}+\lambda N\mathbf{I}_{M}\right)\mathbf{w}=-\mathbf{X}^{T}\left(b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)\\  & \mathbf{w}=-\left(\mathbf{X}^{T}\mathbf{X}+\lambda N\mathbf{I}_{M}\right)^{-1}\mathbf{X}^{T}\left(b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)\end{align*}
$$
故
$$
\mathbf{w}=-\left(\mathbf{X}^{T}\mathbf{X}+\lambda N\mathbf{I}_{M}\right)^{-1}\mathbf{X}^{T}\left(b\left(\mathbf{1}\right)_{N}-\mathbf{y}\right)
$$

---
矩阵$\mathbf{X}^{T}\mathbf{X}+\lambda N\mathbf{I}_{M}$为正定矩阵

这是由于两个矩阵得特征值分别为
$$
\begin{align*} & \lambda_{j}\ge0,j=1,2,\cdots,M\\  & \lambda N>0\end{align*}
$$
其中取$\lambda>0$，而且这是正则化参数，不是矩阵特征值的符号
